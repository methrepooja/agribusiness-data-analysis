##1.Objective
The objective of this project is to collect, explore, clean, and analyze agricultural datasets to identify data quality issues and understand patterns in crop production, agricultural trade, and cereal yield.

## 2. Dataset Description

This project uses three agricultural datasets to study crop production, agricultural trade, and cereal yield.

### Dataset 1: Crop Production
- File: `crop.csv`
- Rows: 38,266
- Columns: 7
- Contains information about state, district, year, season, crop, area, and production.
- This dataset is used to analyze crop production patterns in India.

### Dataset 2: Agricultural Trade
- File: `faostat.csv`
- Rows: 4,122
- Columns: 15
- Contains agricultural import and export information.
- This dataset is used to explore agricultural trade patterns.

### Dataset 3: Cereal Yield
- File: `cereal_yield.csv`
- Rows: 269
- Columns: 70
- Contains cereal yield data measured in kilograms per hectare across countries and years.
- This dataset is used to examine cereal yield trends.

## 3. Data Loading

```python
import pandas as pd

crop = pd.read_csv("crop.csv")
faostat = pd.read_csv("faostat.csv")
cereal_yield = pd.read_csv("cereal_yield.csv")

print("Crop Production:", crop.shape)
print("FAOSTAT:", faostat.shape)
print("Cereal Yield:", cereal_yield.shape)
```

## 4. Dataset Overview

```python
# Display the first 5 rows of each dataset

print("CROP PRODUCTION DATA")
display(crop.head())

print("\nFAOSTAT DATA")
display(faostat.head())

print("\nCEREAL YIELD DATA")
display(cereal_yield.head())
```

### Dataset Information

```python
print("CROP PRODUCTION")
crop.info()

print("\nFAOSTAT")
faostat.info()

print("\nCEREAL YIELD")
cereal_yield.info()
```

### Missing Value Analysis

```python
print("Missing values in Crop Production:")
print(crop.isnull().sum())

print("\nMissing values in FAOSTAT:")
print(faostat.isnull().sum())

print("\nMissing values in Cereal Yield:")
print(cereal_yield.isnull().sum())
```

## 5. Missing Value Analysis

```python
# Check missing values in all datasets

print("CROP PRODUCTION")
print(crop.isnull().sum())

print("\nFAOSTAT")
print(faostat.isnull().sum())

print("\nCEREAL YIELD")
print(cereal_yield.isnull().sum())
```

### Missing Value Percentage

```python
for name, df in {
    "Crop Production": crop,
    "FAOSTAT": faostat,
    "Cereal Yield": cereal_yield
}.items():

    print("\n", name)
    missing = df.isnull().sum()
    percentage = (missing / len(df)) * 100

    result = pd.DataFrame({
        "Missing Values": missing,
        "Percentage": percentage.round(2)
    })

    display(result)
```

## 8. Exploratory Data Analysis

### 8.1 Crop Production by Year

```python
import matplotlib.pyplot as plt

year_production = crop.groupby("Crop_Year")["Production"].sum()

plt.figure(figsize=(10,5))
year_production.plot(marker="o")
plt.title("Total Crop Production by Year")
plt.xlabel("Year")
plt.ylabel("Total Production")
plt.grid(True)
plt.show()
```

### 8.2 Top 10 Crops by Production

```python
top_crops = (
    crop.groupby("Crop")["Production"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))
top_crops.plot(kind="bar")
plt.title("Top 10 Crops by Total Production")
plt.xlabel("Crop")
plt.ylabel("Total Production")
plt.xticks(rotation=45)
plt.show()
```

### 8.3 Crop Production by Season

```python
season_production = crop.groupby("Season")["Production"].sum()

plt.figure(figsize=(8,5))
season_production.plot(kind="bar")
plt.title("Crop Production by Season")
plt.xlabel("Season")
plt.ylabel("Total Production")
plt.xticks(rotation=45)
plt.show()
```

### 8.4 Crop Production by State

```python
top_states = (
    crop.groupby("State_Name")["Production"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,5))
top_states.plot(kind="bar")
plt.title("Top 10 States by Crop Production")
plt.xlabel("State")
plt.ylabel("Total Production")
plt.xticks(rotation=45)
plt.show()
```

### 8.5 Relationship Between Area and Production

```python
plt.figure(figsize=(10, 6))
plt.scatter(crop["Area"], crop["Production"], alpha=0.5, color="teal")
plt.title("Relationship Between Cultivated Area and Crop Production")
plt.xlabel("Area (Hectares)")
plt.ylabel("Production (Tonnes)")
plt.grid(True)
plt.show()
```

## 9. Data Cleaning

### 9.1 Handling Missing Values

```python
# Check missing values before cleaning

print("Missing values before cleaning:")
print(crop.isnull().sum())
```

```python
# Remove rows where Production is missing
crop_clean = crop.dropna(subset=["Production"]).copy()

print("Original rows:", len(crop))
print("Rows after cleaning:", len(crop_clean))
print("Rows removed:", len(crop) - len(crop_clean))
```

### 9.2 Handling Duplicate Records

```python
print("Duplicates before cleaning:", crop_clean.duplicated().sum())

crop_clean = crop_clean.drop_duplicates()

print("Duplicates after cleaning:", crop_clean.duplicated().sum())
```

### 9.3 Final Data Quality Check

```python
print("Final shape:", crop_clean.shape)

print("\nMissing values:")
print(crop_clean.isnull().sum())

print("\nDuplicate rows:")
print(crop_clean.duplicated().sum())
```
